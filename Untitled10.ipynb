```python
from google.colab import drive
from pathlib import Path
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency

# 1. 드라이브 연결
drive.mount('/content/drive')

# 2. CSV가 저장된 폴더 경로 설정
DATA_DIR = Path('/content/drive/MyDrive/type3_week11')
print('설정된 데이터 폴더:', DATA_DIR)

required_files = ['channel_purchase.csv', 'gender_preference.csv']

print('\n[파일 존재 여부 확인]')
for fname in required_files:
    fpath = DATA_DIR / fname
    print(f'{fname}:', 'OK' if fpath.exists() else '없음')

# 3. 데이터 로드 및 탐색
file_path = DATA_DIR / 'channel_purchase.csv'
df = pd.read_csv(file_path)

print('\n[데이터 상위 5행]')
print(df.head())
print('\n[기본 정보]')
print(df.info())

# 4. 교차표 생성 및 카이제곱 검정
ct = pd.crosstab(df['channel'], df['purchase_yn'])
print('\n[교차표]')
print(ct)

chi2, p, dof, expected = chi2_contingency(ct)
expected_df = pd.DataFrame(expected, index=ct.index, columns=ct.columns)

print('\n[검정 결과]')
print('chi-square statistic:', round(chi2, 4))
print('p-value:', round(p, 6))
print('degrees of freedom:', dof)

print('\n[기대도수]')
print(expected_df.round(2))

# 5. 채널별 구매 비율 분석
conversion_rate = pd.crosstab(df['channel'], df['purchase_yn'], normalize='index') * 100
print('\n[채널별 구매 비율(%)]')
print(conversion_rate.round(2))

# 6. 최종 해석 출력
print('\n[최종 해석 출력]')
if p < 0.05:
    print('p-value가 0.05보다 작으므로, 유입채널과 구매여부는 독립이 아니며 서로 관련이 있다고 해석할 수 있습니다.')
else:
    print('p-value가 0.05 이상이므로, 유입채널과 구매여부의 관련성을 확인하기 어렵습니다.')

# 가장 구매 비율이 높은 채널 찾기 (컬럼명이 숫자 1 또는 문자 '1'이거나 마지막 컬럼인 경우 대응)
target_col = 1 if 1 in conversion_rate.columns else ('1' if '1' in conversion_rate.columns else conversion_rate.columns[-1])
best_channel = conversion_rate[target_col].idxmax()
best_rate = conversion_rate[target_col].max()

print(f'구매 비율이 가장 높은 채널은 [{best_channel}]이며, 구매 비율은 {best_rate:.2f}%입니다.')
```

```python
import pandas as pd
from scipy.stats import chi2_contingency

# 1. 데이터 로드 및 탐색
file_path = DATA_DIR / 'gender_preference.csv'
df = pd.read_csv(file_path)

print('[데이터 상위 5행]')
print(df.head())
print('\n[기본 정보]')
print(df.info())

# 2. 교차표 생성
ct = pd.crosstab(df['gender'], df['content_type'])
print('\n[교차표]')
print(ct)

# 3. 카이제곱 검정 수행
chi2, p, dof, expected = chi2_contingency(ct)
expected_df = pd.DataFrame(expected, index=ct.index, columns=ct.columns)

print('\n[검정 결과]')
print('chi-square statistic:', round(chi2, 4))
print('p-value:', round(p, 6))
print('degrees of freedom:', dof)

print('\n[기대도수]')
print(expected_df.round(2))

# 4. 성별 내 콘텐츠 선호 비율 및 최다 선호 추출
ratio = pd.crosstab(df['gender'], df['content_type'], normalize='index') * 100
top_pref = ct.idxmax(axis=1)

print('\n[성별 내 콘텐츠 선호 비율(%)]')
print(ratio.round(2))

print('\n[성별별 최다 선호 콘텐츠]')
print(top_pref)

# 5. 최종 해석 출력
print('\n[최종 해석 출력]')
if p < 0.05:
    print('p-value가 0.05보다 작으므로, 성별에 따라 콘텐츠 선호 분포가 다르다고 해석할 수 있습니다. (성별과 콘텐츠 선호도는 관련이 있음)')
else:
    print('p-value가 0.05 이상이므로, 성별에 따른 콘텐츠 선호 분포 차이를 확인하기 어렵습니다. (성별과 콘텐츠 선호도는 독립적임)')

print() # 한 줄 띄우기
for g in top_pref.index:
    print(f'{g}의 최다 선호 콘텐츠는 [{top_pref[g]}]입니다.')
```
